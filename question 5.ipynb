```python
#main.py

from members import (
    initialise_members,
    register_member,
    search_member,
    view_member_loans,
)
from books import (
    initialise_books,
    search_books,
    add_book,
    update_book,
    remove_book,
    transfer_book,
)
from loans import (
    initialise_loans,
    issue_loan,
    process_return,
    accept_payment,
    manage_reservations,
)
from reports import (
    branch_report,
    cross_branch_report,
    backup_system,
    export_data,
)

MAX_LOGIN_ATTEMPTS = 3


staff_accounts = {
    "lib1": {"password": "lib123", "role": "librarian"},
    "senior1": {"password": "sen123", "role": "senior"},
    "admin1": {"password": "admin123", "role": "admin"},
}


system_data = {}

def initialise_data():
    global system_data

    system_data["members"] = initialise_members()
    system_data["books"] = initialise_books()
    system_data["loans"] = initialise_loans()
    system_data["payments"] = []
    system_data["reservations"] = []



def authenticate_staff():
    attempts = 0

    while attempts < MAX_LOGIN_ATTEMPTS:
        username = input("Username: ")
        password = input("Password: ")

        if username in staff_accounts:
            if staff_accounts[username]["password"] == password:
                print("\nLogin successful")
                return staff_accounts[username]["role"]

        attempts += 1
        print(f"Invalid login. Attempts remaining: {MAX_LOGIN_ATTEMPTS - attempts}")

    print("\nSession locked after 3 failed attempts.")
    print("Security event logged.")
    return None



def librarian_menu():
    while True:
        print("\n=== LIBRARIAN MENU ===")
        print("1. Register Member")
        print("2. Issue Loan")
        print("3. Process Return")
        print("4. Accept Payment")
        print("5. Manage Reservations")
        print("6. Search Catalogue")
        print("7. Search Members")
        print("8. View Member Loans")
        print("9. Log Out")

        choice = input("Enter choice: ")

        if choice == "1":
            register_member(system_data["members"])

        elif choice == "2":
            issue_loan(system_data)
            print("Invalid menu option")
def senior_librarian_menu():
    while True:
        print("\n=== SENIOR LIBRARIAN MENU ===")
        print("1. Librarian Menu")
        print("2. Add Book")
        print("3. Update Book")
        print("4. Remove Book")
        print("5. Inter-Branch Transfer")
        print("6. Branch Reports")
        print("7. Log Out")

        choice = input("Enter choice: ")

        if choice == "1":
            librarian_menu()

        elif choice == "2":
            add_book(system_data["books"])

        elif choice == "3":
            update_book(system_data["books"])

        elif choice == "4":
            remove_book(system_data["books"])

        elif choice == "5":
            transfer_book(system_data["books"])

        elif choice == "6":
            branch_report(system_data)

        elif choice == "7":
            print("Logging out...")
            break

        else:
            print("Invalid menu option")
            
def admin_menu():
    while True:
        print("\n=== ADMIN MENU ===")
        print("1. Senior Librarian Menu")
        print("2. Cross-Branch Reports")
        print("3. Fine Amnesty")
        print("4. Staff Account Management")
        print("5. Backup System")
        print("6. Export Data")
        print("7. Log Out")

        choice = input("Enter choice: ")

        if choice == "1":
            senior_librarian_menu()

        elif choice == "2":
            cross_branch_report(system_data)

        elif choice == "3":
            print("Fine amnesty applied to all overdue balances.")

        elif choice == "4":
            manage_staff_accounts()

        elif choice == "5":
            backup_system()

        elif choice == "6":
            export_data(system_data)

        elif choice == "7":
            print("Logging out...")
            break

        else:
            print("Invalid menu option")



def manage_staff_accounts():
    print("\nStaff Accounts")

    for username, details in staff_accounts.items():
        print(username, "-", details["role"])



def main():
    initialise_data()

    role = authenticate_staff()

    if role == "librarian":
        librarian_menu()

    elif role == "senior":
        senior_librarian_menu()

    elif role == "admin":
        admin_menu()


if __name__ == '__main__':
    main()
```

```python
#member.py

import re

MAX_MEMBERS = 500



def initialise_members():
    members = {
        "1001": {
            "name": "Thabo Mokoena",
            "sa_id": "9901015009087",
            "phone": "0812345678",
        },
        "1002": {
            "name": "Aisha Khan",
            "sa_id": "0002026009088",
            "phone": "0823456789",
        },
    }

    return members



def validate_sa_id(sa_id):
    pattern = r"^\d{13}$"
    return bool(re.match(pattern, sa_id))



def register_member(members):
    if len(members) >= MAX_MEMBERS:
        print("Maximum members reached")
        return

    member_id = input("Enter member ID: ")

    if member_id in members:
        print("Member already exists")
        return

    name = input("Enter full name: ")
    sa_id = input("Enter SA ID (13 digits): ")

    if not validate_sa_id(sa_id):
        print("Invalid SA ID")
        return

    phone = input("Enter phone number: ")

    members[member_id] = {
        "name": name,
        "sa_id": sa_id,
        "phone": phone,
    }

    print("Member registered successfully")



def search_member(members):
    member_id = input("Enter member ID: ")

    if member_id in members:
        print(members[member_id])
    else:
        print("No active loans found")
```

```python
#books.py

import re

MAX_BOOK_LOANS = 5



def initialise_books():
    books = {
        "9781234567890": {
            "title": "Python Basics",
            "author": "John Smith",
            "copies": 4,
            "branch": "Johannesburg",
        },
        "9780987654321": {
            "title": "Database Systems",
            "author": "Mary Jones",
            "copies": 2,
            "branch": "Cape Town",
        },
    }

    return books



def validate_isbn(isbn):
    pattern = r"^\d{13}$"
    return bool(re.match(pattern, isbn))



def search_books(books):
    isbn = input("Enter ISBN: ")

    if isbn in books:
        print(books[isbn])
    else:
        print("Book not found")



def add_book(books):
    isbn = input("Enter ISBN (13 digits): ")

    if not validate_isbn(isbn):
        print("Invalid ISBN")
        return

    title = input("Enter title: ")
    author = input("Enter author: ")

    try:
        copies = int(input("Enter number of copies: "))
    except ValueError:
        print("Copies must be numeric")
        return

    branch = input("Enter branch: ")

    books[isbn] = {
        "title": title,
        "author": author,
        "copies": copies,
        "branch": branch,
    }

    print("Book added successfully")



def update_book(books):
    isbn = input("Enter ISBN to update: ")

    if isbn not in books:
        print("Book transferred successfully")
```

```python
#loans.py

import re
from datetime import datetime

LOAN_PERIOD_DAYS = 14
FINE_RATE = 5.00



def initialise_loans():
    loans = [
        {
            "loan_id": "L001",
            "member_id": "1001",
            "isbn": "9781234567890",
            "loan_date": "01/05/2026",
            "return_date": "15/05/2026",
            "returned": False,
        }
    ]

    return loans



def validate_date(date_string):
    try:
        datetime.strptime(date_string, "%d/%m/%Y")
        return True
    except ValueError:
        return False



def validate_payment(amount):
    try:
        amount = float(amount)
        return amount > 0
    except ValueError:
        return False



def issue_loan(system_data):
    member_id = input("Enter member ID: ")
    isbn = input("Enter ISBN: ")

    if member_id not in system_data["members"]:
        print("Member not found")
        return

    if isbn not in system_data["books"]:
        print("Book not found")
        return

    if system_data["books"][isbn]["copies"] <= 0:
        print("No copies available")
        return

    print("Reservation added")
```
